```python
import pandas as pd
import numpy as np
```

```python
stats2018 = pd.read_json('../webscraper/fantasyStats2018.json')
stats2019 = pd.read_json('../webscraper/fantasyStats2019.json')
```

```python
stats2018.head()
```

```python
stats2019.head()
```

```python
info2019 = pd.DataFrame([stats2019['name'],stats2019['fantasyPoints'],stats2019['pprPoints'],stats2019['draftKingPoints'],stats2019['fanDuelPoints']])
```

```python
                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                            temp_model = pd.merge(stats2018,info2019.transpose(),on='name',how='outer')
```

```python
temp_model
```

```python
temp_model = temp_model.dropna()
```

```python
temp_model.info()
```

```python
temp_model['position'].value_counts()
```

```python
temp_model = temp_model.drop(['teamName','vbd'],axis = 1)
```

```python
position_dummies = pd.get_dummies(temp_model['position'])
```

```python
temp_model = pd.concat([temp_model,position_dummies],axis=1)
```

```python
temp_model = temp_model.drop('position',axis=1)
```

```python
temp_model
```

```python
temp_model['fantasyPoints_y'] = temp_model['fantasyPoints_y'].apply(lambda x: float(x))
temp_model['pprPoints_y'] = temp_model['pprPoints_y'].apply(lambda x: float(x))
temp_model['draftKingPoints_y'] = temp_model['draftKingPoints_y'].apply(lambda x: float(x))
temp_model['fanDuelPoints_y'] = temp_model['fanDuelPoints_y'].apply(lambda x: float(x))
temp_model.info()
```

```python
playerStats2018 = temp_model.drop(['name','fantasyPoints_y','pprPoints_y','draftKingPoints_y','fanDuelPoints_y'],axis=1)
fantasyPoints2019 = temp_model['fantasyPoints_y']
pprPoints2019 = temp_model['pprPoints_y']
draftKingPoints = temp_model['draftKingPoints_y']
fanDuelPoints = temp_model['fanDuelPoints_y']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(playerStats2018, fantasyPoints2019, test_size=0.20, random_state=101)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Activation,Dropout
from tensorflow.keras.constraints import max_norm
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential()
model.add(Dense(32, activation='relu'))

model.add(Dense(16, activation='relu'))

model.add(Dense(8, activation='relu'))

model.add(Dense(1))

model.compile(optimizer='adam',loss='mse')
```

```python
playerStats2018.info()
```

```python
early_stop = EarlyStopping(monitor='val_loss', mode='min', verbose=1, patience=125)
```

```python
model.fit(x=X_train, 
          y=y_train, 
          epochs=1000,
          validation_data=(X_test, y_test), verbose=1,
          callbacks=[early_stop]
          )
```

```python
model_loss = pd.DataFrame(model.history.history)
model_loss.plot()
```

```python
predictions = model.predict_classes(X_test)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import mean_squared_error,mean_absolute_error,explained_variance_score
```

```python
# Our predictions
plt.scatter(y_test,predictions)

# Perfect predictions
plt.plot(y_test,y_test,'r')
```
