```python
temp = stats2016.drop(['fantasyPoints','pprPoints','draftKingPoints','fanDuelPoints'],axis=1)
```

```python
model1 = model1.drop(['teamName','vbd'],axis=1)
```

```python
position_dummies = pd.get_dummies(model1['position'])
position_dummies
```

```python
stats2017[stats2017['name']=='Aaron Rodgers']
```

```python
model1 = pd.concat([model1,position_dummies],axis=1)
```

```python
model1 = model1.drop('position',axis=1)
```

```python
model1 = model1.drop(['QB','RB','TE','WR'],axis=1)
```

```python
model1.head()
```

```python
model1['fantasyPoints'] = model1['fantasyPoints'].apply(lambda x: float(x))
model1['pprPoints'] = model1['pprPoints'].apply(lambda x: float(x))
model1['draftKingPoints'] = model1['draftKingPoints'].apply(lambda x: float(x))
model1['fanDuelPoints'] = model1['fanDuelPoints'].apply(lambda x: float(x))
model1.info()
```

```python
playerStats2016 = model1.drop(['name','fantasyPoints','pprPoints','draftKingPoints','fanDuelPoints'],axis=1)
fantasyPoints2017 = model1['fantasyPoints']
pprPoints2017 = model1['pprPoints']
draftKingPoints = model1['draftKingPoints']
fanDuelPoints = model1['fanDuelPoints']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(playerStats2016, fantasyPoints2017, test_size=0.20, random_state=101)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Activation,Dropout
from tensorflow.keras.constraints import max_norm
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential()
```

```python
model.add(Dense(28, activation='relu'))

model.add(Dense(14, activation='relu'))

model.add(Dense(7, activation='relu'))

model.add(Dense(1))

model.compile(optimizer='adam',loss='mse')
```

```python
early_stop = EarlyStopping(monitor='val_loss', mode='min', verbose=1, patience=100)
```

```python
model.fit(x=X_train, 
          y=y_train, 
          epochs=1000,
          validation_data=(X_test, y_test), verbose=1,
          callbacks=[early_stop]
          )
```

```python
model_loss = pd.DataFrame(model.history.history)
model_loss.plot()
```

```python
predictions = model.predict_classes(X_test)
```
