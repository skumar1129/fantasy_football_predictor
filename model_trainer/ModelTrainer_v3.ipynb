```python
playerStats = temp_model.drop(['name','fantasyPoints_y','pprPoints_y','draftKingPoints_y','fanDuelPoints_y'],axis=1)
fantasyPoints = temp_model['fantasyPoints_y']
pprPoints = temp_model['pprPoints_y']
draftKingPoints = temp_model['draftKingPoints_y']
fanDuelPoints = temp_model['fanDuelPoints_y']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(playerStats, fantasyPoints, test_size=0.30, random_state=101)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Activation,Dropout
from tensorflow.keras.constraints import max_norm
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential()
model.add(Dense(32, activation='relu'))

model.add(Dense(16, activation='relu'))

# model.add(Dense(8, activation='relu'))

model.add(Dense(1))

model.compile(optimizer='adam',loss='mse')
```

```python
early_stop = EarlyStopping(monitor='val_loss', mode='min', verbose=1, patience=100)
```

```python
model.fit(x=X_train, 
          y=y_train, 
          epochs=1000,
          validation_data=(X_test, y_test), verbose=1,
          callbacks=[early_stop]
          )
```

```python
model_loss = pd.DataFrame(model.history.history)
model_loss.plot()
```

```python
predictions = model.predict_classes(X_test)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import mean_squared_error,mean_absolute_error,explained_variance_score
```
